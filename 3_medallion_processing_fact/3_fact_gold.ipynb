```python
%sql
USE CATALOG ecommerce;
```

```python
%sql
CREATE OR REPLACE TABLE gold.gold_order_items AS(
  SELECT 
    TO_CHAR(dt, 'yyyyMMdd') AS date_id,
    dt AS transaction_date,
    order_ts AS transaction_ts,
    order_id,
    customer_id,
    item_seq AS seq_no,
    product_id,
    channel,
    coupon_code,
    CASE 
      WHEN coupon_code IS NOT NULL THEN 1
      ELSE 0
    END AS coupon_flag,
    unit_price_currency,
    quantity,
    unit_price,
    quantity * unit_price AS gross_amount,
    discount_pct AS discount_percent,
    gross_amount * discount_pct AS discount_amount,
    tax_amount,
    gross_amount - discount_amount + tax_amount AS net_amount
  FROM silver.silver_order_items
)
```

```python
%sql
SELECT COUNT(*) FROM gold.gold_order_items
```
