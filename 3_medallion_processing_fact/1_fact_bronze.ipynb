```python
import pyspark.sql.functions as F
from pyspark.sql.types import *
```

```python
source_data_path = '/Volumes/ecommerce/source_data/raw/order_items/landing/*.csv'

schema = StructType([
    StructField('dt', StringType(), True),
    StructField('order_ts', StringType(), True),
    StructField('customer_id', StringType(), True),
    StructField('item_seq', StringType(), True),
    StructField('product_id', StringType(), True),
    StructField('quantity', StringType(), True),
    StructField('unit_price_currency', StringType(), True),
    StructField('unit_price', StringType(), True),
    StructField('discount_pct', StringType(), True),
    StructField('tax_amount', StringType(), True),
    StructField('channel', StringType(), True),
    StructField('coupon_code', StringType(), True),
])

df = spark.read.csv(source_data_path, header=True, schema=schema)

df = df.withColumn('_source_file', F.col('_metadata.file_path'))\
    .withColumn('ingested_at', F.current_timestamp())

display(df.limit(5))
```

```python
df.count()
```

```python
df.write.mode('overwrite')\
    .format('delta')\
    .option('mergeSchema', True)\
    .saveAsTable('ecommerce.bronze.bronze_order_items')
```
